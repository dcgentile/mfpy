```python
from mfpy.utils import label_series
from mfpy.clustering import SegmentBasedCluster
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

filename = "../../../data/VDP/V_3D_1ps_shifted.dat"
data = np.loadtxt(filename)
phi = data[:,1].reshape(-1, 1)
psi = data[:,2].reshape(-1, 1)
chi = data[:,3].reshape(-1, 1)
```

```python
# T = data.shape[0]
T = 10000
l = 2*T
u = 3*T
x = psi[l:u]
w = 32
q = 0.75
sbc = SegmentBasedCluster(w, q)
params, score = sbc.gridsearch(x)
#sbc = SegmentBasedCluster(w, q, periodic=False, method="spectral", num_clusters=2)
```

```python
sbc = SegmentBasedCluster(params["window_size"], params["quantile"])
sbc.fit(x)
```

```python
segment_labels = sbc.segment_labels_
labels = sbc.point_labels_
changes = np.where(sbc.change_points_ !=0)[0] + l

df = pd.DataFrame({"time": range(l,u), "value": x[:,0], "label": labels})

score = sbc.score(x, 100)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="time", y="value", hue="label", palette="bright", s = 5)
plt.vlines(changes, ymin=plt.ylim()[0], ymax=plt.ylim()[1], colors='red', alpha=0.5, zorder=0)
plt.title(f"Score: {score:.3f}")
plt.show()

n_clusters = len(df["label"].unique())

g = sns.FacetGrid(df, col="label", col_wrap=3, height=4)
g.map(sns.histplot, "value", kde=True, stat="probability")
g.set_axis_labels("Value", "Frequency")
g.set_titles("Cluster {col_name}")
data_min, data_max = df['value'].min(), df['value'].max()
xticks = np.linspace(data_min, data_max, 8)  # 8 evenly spaced ticks

# Apply ticks and labels to ALL subplots
for ax in g.axes.flat:
    ax.set_xticks(xticks)
    ax.set_xticklabels([f'{x:.1f}' for x in xticks])  # format labels to 1 decimal
    ax.tick_params(labelbottom=True)

plt.show()
```
