```python
from mfpy.utils import label_series
from mfpy.clustering import SegmentBasedCluster
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

T=20000

adp_filename = "../../../data/ADP/A_2D_1ps.dat"
adp_data = np.loadtxt(adp_filename)[:T,:]
vdp_filename = "../../../data/VDP/V_3D_1ps.dat"
vdp_data = np.loadtxt(vdp_filename)[:T,1:]
```

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

fig = plt.figure(figsize=(12, 8))
gs = GridSpec(2, 6, figure=fig)

# Top row: 2 plots centered (each spans 3 columns)
ax0 = fig.add_subplot(gs[0, 0:3])
ax1 = fig.add_subplot(gs[0, 3:6], sharey=ax0)

# Bottom row: 3 plots (each spans 2 columns)
ax2 = fig.add_subplot(gs[1, 0:2])
ax3 = fig.add_subplot(gs[1, 2:4], sharey=ax2)
ax4 = fig.add_subplot(gs[1, 4:6], sharey=ax2)

ticks = [-180, -120, -60, 0, 60, 120, 180]

s=0.1

# Plot adp_data (top row)
ax0.scatter(range(len(adp_data)), adp_data[:, 0], s=s, rasterized=True)
ax1.scatter(range(len(adp_data)), adp_data[:, 1], s=s, rasterized=True)

# Plot vdp_data (bottom row)
ax2.scatter(range(len(vdp_data)), vdp_data[:, 0], s=s, rasterized=True)
ax3.scatter(range(len(vdp_data)), vdp_data[:, 1], s=s, rasterized=True)
ax4.scatter(range(len(vdp_data)), vdp_data[:, 2], s=s, rasterized=True)

# Set yticks
ax0.set_yticks(ticks)
ax2.set_yticks(ticks)

# Set ylabels
ax0.set_ylabel(r"ADP $\phi$")
ax1.set_ylabel(r"ADP $\psi$")
ax2.set_ylabel(r"VDP $\phi$")
ax3.set_ylabel(r"VDP $\psi$")
ax4.set_ylabel(r"VDP $\chi$")


plt.tight_layout()
plt.savefig("angle-compare.pdf", bbox_inches="tight")
```

```python
data_scaled = data_subset * 360 - 180
labels = sbc.point_labels_
N = 6  # Choose number of clusters to plot
unique_labels, counts = np.unique(labels, return_counts=True)
top_labels = unique_labels[np.argsort(counts)[-N:]][::-1]
mask = np.isin(labels, top_labels)
fig = plt.figure()
ticks = [-180, -90, 0, 90, 180]
ax = fig.add_subplot()
scatter = ax.scatter(data_scaled[mask, 0], data_scaled[mask, 1],
                     c=labels[mask], cmap='tab10',s=.1, rasterized=True)
plt.colorbar(scatter, label='Cluster')
ax.set_xticks(ticks)
ax.set_yticks(ticks)
plt.savefig("adp-clusters.pdf", bbox_inches="tight")
plt.show()
```

```python
labels = sbc.point_labels_
unique_labels, counts = np.unique(labels, return_counts=True)
top_labels = unique_labels[np.argsort(counts)[-N:]][::-1]
cmap = plt.cm.tab10
cols = 3
rows = int(np.ceil(N / cols))
fig = plt.figure(figsize=(5*cols, 5*rows))


for i, label in enumerate(top_labels):
    ax = fig.add_subplot(rows, cols, i+1)
    mask = labels == label
    ax.scatter(data_scaled[mask, 0], data_scaled[mask, 1],
               c=[cmap(label)], s=.1, rasterized=True)
    ax.set_title(f'Cluster {label} (n={np.sum(mask)})')
    ax.set_xticks(ticks)
    ax.set_yticks(ticks)
plt.savefig("adp-top-individual-clusters.pdf", bbox_inches="tight")
plt.show()
```
